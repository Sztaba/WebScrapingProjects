```python
from bs4 import BeautifulSoup
import json
import requests
```

```python
with open('source.html', 'r', encoding='utf-8') as f:
    games = f.read()
```

```python
soup = BeautifulSoup(games, 'html.parser')

url = "https://steamdb.info/"

games_by_date = {}

for element in soup.find_all('div', class_='calendar-day'):
    if "calendar-empty" not in element.get('class'):
        for app in element.find_all('a'):
            print(url + app.get('href') + "charts")
            print(element.find('div', class_='calendar-day-badge').text)
            response = requests.get(url + app.get('href') + "charts")
            print(response, '\n')
    # splitted_date = tr.find_all('td')[6].text.strip().split()
    # date = splitted_date[len(splitted_date) - 1]
    
    # if date in games_by_date:
    #     games_by_date[date] += 1
    # else:
    #     games_by_date[date] = 1
```

```python
# save games dictionary to json
with open('Data/open_world_games_only.json', 'w', encoding='utf-8') as f:
    json.dump(games_by_date, f)
```
