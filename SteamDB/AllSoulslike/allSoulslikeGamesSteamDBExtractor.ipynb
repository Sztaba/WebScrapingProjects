```python
from bs4 import BeautifulSoup
import json
```

```python
with open('Data/Tags_Souls-like_SteamDB_games_only.html', 'r', encoding='utf-8') as f:
    soulsGames = f.read()
```

```python
with open('Data/Tags_Souls-like_SteamDB_limit.html', 'r', encoding='utf-8') as f:
    soulsGames = f.read()
```

```python
soup = BeautifulSoup(soulsGames, 'html.parser')

games_by_date = {}

for tr in soup.find('tbody').find_all('tr', class_='app'):
    splitted_date = tr.find_all('td')[6].text.strip().split()
    date = splitted_date[len(splitted_date) - 1]
    
    if date in games_by_date:
        games_by_date[date] += 1
    else:
        games_by_date[date] = 1
```

```python
# save games dictionary to json
with open('Data/souls_like_games_only.json', 'w', encoding='utf-8') as f:
    json.dump(games_by_date, f)
```

```python
# save games dictionary to json
with open('Data/souls_like_limit.json', 'w', encoding='utf-8') as f:
    json.dump(games_by_date, f)
```
