```python
from bs4 import BeautifulSoup
import os
import re
```

```python
def getHtmlText(file):
    html = file.read()
    return BeautifulSoup(html, 'html.parser').text

def extractAll(folder_path, target_folder_path):
    file_text = ""
    for file_name in os.listdir(folder_path):
        with open(os.path.join(folder_path, file_name), 'r', encoding='utf-8') as f:
            print(f"{file_name} OPENED")
            file_text = getHtmlText(f)
        cleaned_text = re.sub(r'\n{3,}', '\n\n', file_text)
        with open(os.path.join(target_folder_path, file_name.replace(".html", ".txt")), 'w', encoding='utf-8') as f:
            f.write(cleaned_text)

reviews_folder_path = "data/AC Shadows/htmls"
target_folder_path = "data/AC Shadows/html texts"
extractAll(reviews_folder_path, target_folder_path)
```
