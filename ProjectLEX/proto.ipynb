```python
import os
from collections import Counter
from nltk import ngrams
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
import string
import nltk
import spacy
from nltk.stem import WordNetLemmatizer
from nltk import pos_tag
from nltk.corpus import wordnet
```

```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('averaged_perceptron_tagger')
nlp = spacy.load("en_core_web_sm")
lemmatizer = WordNetLemmatizer()
```

```python
def load_reviews(folder_path):
    reviews = []
    for filename in os.listdir(folder_path):
        if filename.endswith(".txt"):
            with open(os.path.join(folder_path, filename), 'r', encoding='utf-8') as f:
                reviews.append(f.read())
    return reviews

def load_review(file_path):
    with open(file_path, 'r', encoding='utf-8') as f:
        return f.read()

def preprocess_spacy(text):
    doc = nlp(text.lower())
    tokens = [token.lemma_ for token in doc if not token.is_punct and not token.is_stop]
    return tokens

def extract_ngrams_spacy(reviews, n):
    all_ngrams = []
    for review in reviews:
        tokens = preprocess_spacy(review)
        all_ngrams.extend(ngrams(tokens, n))
    return Counter(all_ngrams)

def get_wordnet_pos(treebank_tag):
    if treebank_tag.startswith('J'):
        return wordnet.ADJ
    elif treebank_tag.startswith('V'):
        return wordnet.VERB
    elif treebank_tag.startswith('N'):
        return wordnet.NOUN
    elif treebank_tag.startswith('R'):
        return wordnet.ADV
    else:
        return wordnet.NOUN

def preprocess_nltk(text):
    tokens = word_tokenize(text.lower())
    tagged = pos_tag(tokens)
    lemmatized = [
        lemmatizer.lemmatize(word, get_wordnet_pos(pos))
        for word, pos in tagged
        if word not in string.punctuation and word not in stopwords.words('english')
    ]
    return lemmatized

# OLD
def preprocess(text):
    tokens = word_tokenize(text.lower())
    tokens = [t for t in tokens if t not in string.punctuation and t not in stopwords.words('english')]
    return tokens

def extract_ngrams_nltk(reviews, n):
    all_ngrams = []
    for review in reviews:
        tokens = preprocess_nltk(review)
        all_ngrams.extend(ngrams(tokens, n))
    return Counter(all_ngrams)
```

```python
print(preprocess(load_review("data/Dragon Age The Veilguard/Eurogamer.txt")))
print(preprocess_spacy(load_review("data/Dragon Age The Veilguard/Eurogamer.txt")))
print(preprocess_nltk(load_review("data/Dragon Age The Veilguard/Eurogamer.txt")))
```

```python
# Ścieżka do folderu z recenzjami
folder_path = 'data/Dragon Age The Veilguard'

reviews = load_reviews(folder_path)
print("---------REVIEWS---------")
print(reviews)
```

```python
bigrams_nltk = extract_ngrams_nltk(reviews, 2)
print("\n\n---------BIRAMS---------")
print(bigrams_nltk)
trigrams_nltk = extract_ngrams_nltk(reviews, 3)
print("\n\n---------TRIGRAMS---------")
print(trigrams_nltk)
```

```python
bigrams_spacy = extract_ngrams_spacy(reviews, 2)
print("\n\n---------BIRAMS---------")
print(bigrams_spacy)
trigrams_spacy = extract_ngrams_spacy(reviews, 3)
print("\n\n---------TRIGRAMS---------")
print(trigrams_spacy)
```

```python
print("\nTop 20 bigramów NLTK:")
for ngram, count in bigrams_nltk.most_common(100):
    print(f"{ngram}: {count}")

print("\nTop 20 trigramów NLTK:")
for ngram, count in trigrams_nltk.most_common(40):
    print(f"{ngram}: {count}")
```
