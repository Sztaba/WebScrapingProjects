```python
import os
from collections import Counter
from nltk import ngrams
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
import string
import nltk
import spacy
from nltk.stem import WordNetLemmatizer
from nltk import pos_tag
from nltk.corpus import wordnet
```

```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('averaged_perceptron_tagger')
nlp = spacy.load("en_core_web_sm")
lemmatizer = WordNetLemmatizer()
```

```python
def load_reviews(folder_path):
    reviews = []
    for filename in os.listdir(folder_path):
        if filename.endswith(".txt"):
            with open(os.path.join(folder_path, filename), 'r', encoding='utf-8') as f:
                reviews.append(f.read())
    return reviews

def load_review(file_path):
    with open(file_path, 'r', encoding='utf-8') as f:
        return f.read()
    


    

def get_wordnet_pos(treebank_tag):
    if treebank_tag.startswith('J'):
        return wordnet.ADJ
    elif treebank_tag.startswith('V'):
        return wordnet.VERB
    elif treebank_tag.startswith('N'):
        return wordnet.NOUN
    elif treebank_tag.startswith('R'):
        return wordnet.ADV
    else:
        return wordnet.NOUN

def preprocess_nltk(text):
    tokens = word_tokenize(text.lower())
    tagged = pos_tag(tokens)
    lemmatized = [
        lemmatizer.lemmatize(word, get_wordnet_pos(pos))
        for word, pos in tagged
        if word not in string.punctuation and word not in stopwords.words('english')
    ]
    return lemmatized

def extract_ngrams_nltk(reviews, n):
    all_ngrams = []
    for review in reviews:
        tokens = preprocess_nltk(review)
        all_ngrams.extend(ngrams(tokens, n))
    return Counter(all_ngrams)
```

```python
from sklearn.feature_extraction.text import CountVectorizer

preprocessed_reviews = [preprocess_nltk(r) for r in reviews]
joined_reviews = [' '.join(tokens) for tokens in preprocessed_reviews]
vectorizer = CountVectorizer(ngram_range=(3, 5), min_df=2)  # pokaże 3-5gramy, które występują w wielu recenzjach
X = vectorizer.fit_transform(joined_reviews)
phrases = vectorizer.get_feature_names_out()
print(vectorizer)
print(X)
print(phrases)
```

```python
import numpy as np

# X is your sparse matrix from CountVectorizer
# Sum over rows (axis=0) to get total count per column (phrase)
total_phrase_counts = np.asarray(X.sum(axis=0)).flatten()

# Get the actual phrases
phrases = vectorizer.get_feature_names_out()

# Combine into a list of (phrase, count), and sort by count descending
phrase_counts = sorted(zip(phrases, total_phrase_counts), key=lambda x: x[1], reverse=True)

# Example: print top 10 most frequent phrases
for phrase, count in phrase_counts:
    print(f"{phrase}: {count}")
```

```python
from collections import defaultdict

# Convert to COOrdinate format to iterate efficiently
X_coo = X.tocoo()

# Create dictionary: phrase_index → {review_index → count}
phrase_distribution = defaultdict(lambda: defaultdict(int))

for row, col, count in zip(X_coo.row, X_coo.col, X_coo.data):
    phrase_distribution[col][row] = count  # col = phrase index, row = review index

# Optional: map index to actual phrase
phrase_dist_human = {
    phrases[col]: dict(review_counts)
    for col, review_counts in phrase_distribution.items()
}

# Example: print distribution for a specific phrase
example_phrase = "feel like bioware"
if example_phrase in phrase_dist_human:
    print(f"Distribution for '{example_phrase}':")
    for review_idx, count in phrase_dist_human[example_phrase].items():
        review_names = os.listdir("data/AC Shadows/html texts")
        print(f" - Review {review_names[review_idx]}: {count} times")

import pandas as pd

# Build rows: phrase, review_idx, count
rows = []
for phrase_idx, review_counts in phrase_distribution.items():
    phrase = phrases[phrase_idx]
    for review_idx, count in review_counts.items():
        review_names = os.listdir("data/AC Shadows/html texts")
        rows.append((phrase, review_names[review_idx].replace(".txt", ""), count))

df = pd.DataFrame(rows, columns=["Phrase", "Review", "Count"])
print(df)
df.to_csv("data/AC Shadows/df.csv")

```
